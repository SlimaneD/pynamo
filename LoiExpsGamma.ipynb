# **Lois exponentielles générant une loi Gamma**

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import gamma
import ipywidgets as widgets
from ipywidgets import interact, IntSlider, FloatSlider

# Activation du rendu LaTeX dans Matplotlib
plt.rcParams['text.usetex'] = True
plt.rcParams['font.size'] = 14

def somme_exponentielles_vs_gamma(n, lam=1.0, N=50000):
    """
    Compare la somme de n exponentielles Exp(λ) simulées
    avec la densité théorique de la loi Gamma(n, λ).
    """

    # Simulations
    samples = np.random.exponential(scale=1/lam, size=(N, n))
    sums = samples.sum(axis=1)

    # Densité théorique Gamma
    x = np.linspace(0, np.percentile(sums, 99.5), 400)
    pdf_gamma = gamma.pdf(x, a=n, scale=1/lam)

    plt.figure(figsize=(8, 5))

    # Histogramme des sommes simulées
    plt.hist(
        sums,
        bins=60,
        density=True,
        alpha=0.4,
        label=rf"$\mathrm{{Somme\ simulee}},\ n={n}$"
    )

    # Courbe Gamma théorique
    plt.plot(
        x,
        pdf_gamma,
        linewidth=2,
        label=rf"$\Gamma(\alpha={n},\,\lambda={lam})$"
    )

    # Labels en LaTeX
    plt.xlabel(r"$x$")
    plt.ylabel(r"$f(x)$")
    plt.title(
        rf"Somme de $n={n}$ exponentielles $\mathrm{{Exp}}(\lambda={lam})$ "
        rf"vs $\Gamma({n},\,\lambda={lam})$"
    )

    plt.grid(False)
    plt.legend()
    plt.show()


# --- Widgets interactifs ---
interact(
    somme_exponentielles_vs_gamma,
    n=IntSlider(min=1, max=150, step=1, value=1, description="n"),
    lam=FloatSlider(min=0.2, max=2.0, step=0.2, value=1.0, description=r"$\lambda$"),
    N=IntSlider(min=5000, max=100000, step=5000, value=50000, description="N sims")
);
```
